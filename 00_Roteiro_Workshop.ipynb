# Mini-Curso: Databricks para Desenvolvedores (V3 - Unity Catalog)
**Roteiro Estruturado e Arquitetural | BRQ Digital Solutions**

## Pré-requisitos do Lab
* **Infraestrutura Azure:** Conta ativa com créditos e Azure SQL Database instanciado (`AdventureWorks_LT`).
* **Segurança:** Azure Key Vault configurado contendo os secrets `sql-user` e `sql-password`.
* **Databricks Workspace:** Workspace Premium (essencial para habilitar o Unity Catalog) e Secret Scope vinculado ao Key Vault criado.
* **Unity Catalog:** Metastore configurado e catálogos/schemas criados (ex: `dev_catalog.bronze`).
* **Conhecimento Prévio:** Lógica de programação orientada a objetos (Python) e familiaridade básica com SQL.

## Diagramação Lógica do Fluxo (Overview)
Explique o fluxo arquitetural antes do código:
1. O **Databricks Cluster** (Compute) solicita as credenciais ao Secret Scope.
2. O **Secret Scope** resolve o acesso transparente via Azure Key Vault.
3. O driver do **PySpark (via JDBC)** abre conexão com o Azure SQL Database, executa o Predicate Pushdown e extrai os dados.
4. O DataFrame em memória é processado e persistido no **Unity Catalog** usando a nomenclatura lógica (`catalog.schema.table`), abstraindo a URI física do armazenamento (Datalake).
